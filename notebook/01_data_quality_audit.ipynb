# Phase 1 — Data Quality Audit

## Objective

This notebook validates the processed Criteo Uplift Prediction Dataset before causal analysis.

The audit covers:

- File accessibility
- Dataset dimensions
- Column names and data types
- Sample observations
- Missing values
- Binary-variable validity
- Treatment allocation
- Treatment-control feature balance
- Logical consistency between treatment and exposure

## Data Source

Criteo Uplift Prediction Dataset v2.1, stored locally as a Parquet file.

```python
from pathlib import Path

import duckdb
import pandas as pd
from IPython.display import display


def find_project_root(start_path: Path) -> Path:
    """Find the project root by looking for the data and src folders."""
    candidates = [start_path, *start_path.parents]

    for candidate in candidates:
        if (candidate / "data").exists() and (candidate / "src").exists():
            return candidate

    raise FileNotFoundError("Could not locate the project root.")


PROJECT_ROOT = find_project_root(Path.cwd().resolve())

PARQUET_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "criteo-uplift-v2.1.parquet"
)

if not PARQUET_PATH.exists():
    raise FileNotFoundError(f"Parquet file not found: {PARQUET_PATH}")

PARQUET_SQL_PATH = str(PARQUET_PATH).replace("'", "''")

connection = duckdb.connect()

print(f"Project root: {PROJECT_ROOT}")
print(f"Parquet file: {PARQUET_PATH}")
print("Database connection ready.")
```
